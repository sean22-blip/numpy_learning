# NumPy Shape Prediction Practice (Days 1-2)

**How to use:**
1. Run the setup cell below.
2. In each level, replace `None` with your guess (a tuple like `(3, 4)`, an int, or a dtype string for Level 6).
3. Run the cell. You get ✅ / ❌ per question, and a score.
4. Wrong? Fix your guess and re-run. Stuck? Use `check(level, start=..., reveal=True)` to see the answer.

Rules of thumb:
- An integer index removes an axis; a slice keeps it.
- `.T` reverses all axes.
- `-1` in reshape means "infer this dimension".
- `arange` excludes the stop; `linspace` includes it by default.
- Shape `()` is a 0-d scalar array.

```python
import numpy as np

def check(level, start=1, reveal=False):
    """Compare your guesses against the real result of each expression."""
    correct = answered = 0
    for i, (expr, guess) in enumerate(level, start):
        actual = eval(expr)
        if guess is None:
            print(f"{i:>2}. ⬜ not answered   {expr}")
            continue
        answered += 1
        if isinstance(actual, np.dtype):
            try:
                ok = np.dtype(guess) == actual
            except TypeError:
                ok = False
        else:
            ok = guess == actual
        correct += ok
        if ok:
            print(f"{i:>2}. ✅ {expr}  ->  {actual}")
        elif reveal:
            print(f"{i:>2}. ❌ {expr}  you said {guess!r}, actual {actual}")
        else:
            print(f"{i:>2}. ❌ {expr}  you said {guess!r}  (try again)")
    print(f"\nScore: {correct}/{len(level)}  ({answered} answered)")
```

## Level 1: Basics

```python
# Replace each None with your guess
level_1 = [
    ("np.zeros(5).shape", None),
    ("np.ones((3, 4)).shape", None),
    ("np.zeros((2, 3, 4)).shape", None),
    ("np.arange(10).shape", None),
    ("np.arange(2, 10).shape", None),
    ("np.arange(0, 10, 2).shape", None),
    ("np.linspace(0, 1, 5).shape", None),
    ("np.array([1, 2, 3]).shape", None),
    ("np.array([[1, 2, 3]]).shape", None),
    ("np.array([[1], [2], [3]]).shape", None),
]

check(level_1, start=1)
```
